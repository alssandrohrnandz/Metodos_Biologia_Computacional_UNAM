```python
#Ejercicio, integra else y elif para evaluar se la secuncia mide 3, 4 o 5 pdb
seq_nueva='CGTGC'
```

```python
def ContenidoGC(DNA):
  counter=0 #es una variable a la cual le asignamos un valor, si no lo ponemos estamos creando una variable que no existe
  if len(DNA)== 3:#es el primer if
   if (DNA)[0]== 'C' or DNA[0]=='G':
    counter= counter+1
   if (DNA)[1]== 'C' or DNA[1]=='G':
    counter= counter+1
   if (DNA)[2]== 'C' or DNA[2]=='G':
    counter+1
   return counter/3.0
  elif len(DNA)==4:
   if (DNA)[0]== 'C' or DNA[0]=='G':
    counter=counter+1
   if (DNA)[1]== 'C' or DNA[1]=='G':
    counter= counter+1
   if (DNA)[2]== 'C' or DNA[2]=='G':
    counter=counter+1
   if (DNA)[3]== 'C' or DNA[3]=='G':
    counter=counter+1
   return counter/4.0
  elif len(DNA)== 5:
    if (DNA)[0]=='c' or DNA[0]=='G':
      counter=counter+1
    if (DNA)[1]=='c' or DNA[1]=='G':
      counter=counter+1
    if (DNA)[2]=='c' or DNA[2]=='G':
      counter=counter+1
    if (DNA)[3]=='c' or DNA[3]=='G':
      counter=counter+1
    return counter/5.0
```

```python
#Loops
seq_5='AGCTCTGACTAGACTATAGCATCTATACGATCGCTCTCTCGATGCGCCT'
```

```python
#lA FUNCIÓN 'for' va a evaluar sobre cada posición de la secuencia
def ContenidoGC_v2(DNA):
  counter=0
  for base in DNA: #Va revisando base por base
   if base == 'C' or base =='G':
    counter = counter +1
  return counter/len(DNA)
```

```python
ContenidoGC_v2(seq_5)
```
