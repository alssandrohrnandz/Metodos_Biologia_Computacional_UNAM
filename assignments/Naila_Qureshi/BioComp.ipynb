Ejercicio: Integra else y elif para evaluar si tu secuencia mide de tres, cuatro o cinco pares de bases.

```python
seq = "AUGCA"
```

```python
def ContentidoGC(DNA):
  counter = 0
  if len(DNA) == 3: #Este es el primer if
    if DNA[0] == "C" or DNA[0] == "G":
      counter = counter + 1
    if DNA[1] == "C" or DNA[1] == "G":
      counter = counter + 1
    if DNA[2] == "C" or DNA[2] == "G":
      counter = counter + 1
    return counter/3.0
  elif len(DNA) == 4: #Aquí integramos elif
    if DNA[0] == "C" or DNA[0] == "G":
      counter = counter + 1
    if DNA[1] == "C" or DNA[1] == "G":
      counter = counter + 1
    if DNA[2] == "C" or DNA[2] == "G":
      counter = counter + 1
    if DNA[3] == "C" or DNA[3] == "G":
      counter = counter + 1
    return counter/4.0
  elif len(DNA) == 5:
    if DNA[0] == "C" or DNA[0] == "G":
      counter = counter + 1
    if DNA[1] == "C" or DNA[1] == "G":
      counter = counter + 1
    if DNA[2] == "C" or DNA[2] == "G":
      counter = counter + 1
    if DNA[3] == "C" or DNA[3] == "G":
      counter = counter + 1
    if DNA[4] == "C" or DNA[4] == "G":
      counter = counter + 1
    return counter/5.0
  else:
    print("La secuencia no mide entre 3 y 5 pares de bases")
```

```python
#DNA = seq
ContentidoGC(seq)
```

```python
seqlarga = "TATATATTATATATTTTAAATATTTAUCGGCUCGAGAGGCUAGUCGAUCGAUGCUACGUCGUCGAUCGAUCGAUCGAUGCUAGCUACGUAGCUACGUAGCUASGCUAGCUAGCUACGUACGAUAAAAAAAAAAAAAAAAA"
```

```python
#La función FOR evalua sobre cada posición de la secuencia (te ahorra el choro del ejercicio anterior)
def ContentidoGC_2(DNA):
  counter = 0
  for base in DNA:
    if base == "C" or base == "G":
      counter = counter + 1
  return counter/len(DNA)
```

```python
ContentidoGC_2(seqlarga)
```

```python
from Bio import Entrez #Estamos llamando a la paquetería de biopython (Entrez es una paquetería del NCBI)
accesion_id = "NC_012920" #Este es el número de acceso de gen. En este caso es del genoma mitocondrial humano
handle = Entrez.efetch(
    db = "nucleotide",
    id = accesion_id,
    rettype = "fasta",
    retmote = "text"
)
seq_test = handle.read()
handle.close()
print(seq_test)
```
