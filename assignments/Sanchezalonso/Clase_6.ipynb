```python
# Este es el llamado de los módulos y bases de datos en python
import pandas as pd
from sklearn.datasets import load_iris
```

```python
#self<- todas las variables que le metamos
nombre= 'Juan'
organismo='Homo sapeins'
seq_1='agatagcagacagcat'
```

```python
#para poder generar un objeto nosotros necesitamos una clase
class Genoma:
  def __init__(self, nombre, organismo, secuencia):
    self.nombre = nombre
    self.organismo= organismo
    self.secuencia= secuencia

  def a_diccionario(self):
    return{
      "nombre": self.nombre,
      "organismo": self.organismo,
      "longitud": len(self.secuencia),
      "contenidoGC": (self.secuencia.count('C') + self.secuencia.count('G') / len(self.secuencia))
     }
```

```python
g1=Genoma('Juan', 'Homo sapiens', 'ATCTATCTGGTGATGC')
```

```python
genomas = [
    Genoma('ID001', 'Homo sapiens', 'AGTCGTGCGTAAATGC'),
    Genoma('ID002', 'Homo sapiens', 'AGTCGGTGCGAATAAATGC'),
    Genoma('ID003', 'Homo sapiens', 'AATCGTGCGTAAATGC'),
    Genoma('ID004', 'Homo neandethalensis', 'AGTCAGATAGAGGCGTAAATGC'),
    Genoma('ID005', 'Homo denisova', 'AGTCAGATAGAGGCGATAGTAGC'),
    Genoma('ID006', 'Homo erectus', 'AGTCAGTGATGGGCGTAAATGC'),
    Genoma('ID007', 'Homo erectus', 'AGTCAGATAGAGAAATTGAAATGC'),
    Genoma('ID008', 'Homo denisova', 'AGTCAGATAGAGGCGTAGCAATGC'),
    Genoma('ID009', 'Homo denisova', 'AGTCAGATAGAGGCGTAGTTGC'),
    Genoma('ID0010', 'Homo erectus', 'AGTCAGATAGAGGCGAGTACTGC'),
    Genoma('ID0011', 'Homo neandethalensis', 'AGTCAGATAGAGAGTACTAAATGC'),
]
```

```python
# Esta es la función para transformar bases a DataFramen
df = pd.DataFrame([g.a_diccionario() for g in genomas])
```

```python
# ahora vamos a sacr los rangos minimos y máximos de los petalos y sepalos
df.groupby('especie')['petalo_largo'].agg(['min', 'max'])
```

```python
df.groupby('especie')['sepalo_largo'].agg(['min', 'max'])
```

```python
# queremos explorar los valores atipicos de cada columna usando el rango intercuantil
for columna in ['sepalo_largo','sepalo_ancho', 'petalo_largo', 'petalo_ancho']:
  q1 = df[columna].quantile(0.25)
  q3 = df[columna].quantile(0.75)
  iqr = q3-q1
  atipicos = df[(df[columna]< q1 - 1.5 * iqr) | (df[columna] > q3 + 1.5 *iqr)]
  print(f'{columna}:{len(atipicos)} valores atipicos')
```
