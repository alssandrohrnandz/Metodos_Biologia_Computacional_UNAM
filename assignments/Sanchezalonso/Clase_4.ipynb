```python
# Actividad pedir a la IA que les genere un código que descargue una secuencia real de NCBI que genere una funcion que calcule la cantidad de codones de paro (AUG, ATG), la ejecuten. copien y peguen en el chat.
import urllib.request

def descargar_secuencia_ncbi(accession_id: str) -> str:
    """
    # Descarga una secuencia en formato FASTA desde NCBI usando e-utilities.
    """
    url = f"https://eutils.ncbi.nlm.nih.gov/entrez/eutils/efetch.fcgi?db=nuccore&id={accession_id}&rettype=fasta&retmode=text"

    try:
        with urllib.request.urlopen(url) as response:
            lineas = response.read().decode('utf-8').splitlines()
            # Omitir el encabezado de la secuencia FASTA (línea que empieza con '>')
            secuencia = "".join(linea.strip() for linea in lineas if not linea.startswith('>'))
            return secuencia.upper()
    except Exception as e:
        print(f"Error al descargar la secuencia: {e}")
        return ""

def contar_codones(secuencia: str, codones_objetivo: list) -> dict:
    """
    Cuenta la cantidad de ocurrencias de una lista de codones en el marco de lectura principal (de 3 en 3).
    """
    conteo = {codon: 0 for codon in codones_objetivo}

    # Recorrer la secuencia de 3 en 3 nucleótidos
    for i in range(0, len(secuencia) - 2, 3):
        triplete = secuencia[i:i+3]
        if triplete in conteo:
            conteo[triplete] += 1

    return conteo

# --- Ejemplo de uso ---
# Usamos un ID de accesión de NCBI (ejemplo: NM_000518.5 - Beta-globina humana)
accession_id = "NM_000518.5"
secuencia_dna = descargar_secuencia_ncbi(accession_id)

if secuencia_dna:
    print(f"Secuencia {accession_id} descargada exitosamente. Longitud: {len(secuencia_dna)} nt.")

    # Codones de inicio (ATG en ADN)
    codones_inicio = ["ATG"]
    conteo_inicio = contar_codones(secuencia_dna, codones_inicio)

    # Codones de paro estándar en ADN (TAA, TAG, TGA)
    codones_paro = ["TAA", "TAG", "TGA"]
    conteo_paro = contar_codones(secuencia_dna, codones_paro)

    print("\n--- Resultados del análisis ---")
    print(f"Codones de Inicio (ATG): {conteo_inicio}")
    print(f"Codones de Paro (TAA, TAG, TGA): {conteo_paro}")
    print(f"Total de codones de paro: {sum(conteo_paro.values())}")
```
