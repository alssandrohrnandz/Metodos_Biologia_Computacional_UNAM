```python
from enum import nonmember
#Podemos agregar métodos que vivan dentro de la clase
class Genoma:
  def __init__ (self, nombre, secuencia):
    self.nombre = nombre
    self.secuencia = secuencia

  def longitud(self):
    return len(self.secuencia)

  def ContenidoGC(self):
    counter = 0
    secuencia = self.secuencia.upper()
    for base in secuencia:
      if base == 'G' or base == 'C':
         counter = counter + 1
    return counter / len(self.secuencia)
```

```python
g1 = Genoma('Homo sapiens', 'AGTACTAGTAGTAGACAGACACGATGCAGTACAGTACGATCCGGTG')
```

```python
print(f" La secuencia con nombre {g1.nombre} tiene una longitud total de {g1.longitud()}pb y tenemos un contenido GC de {g1.ContenidoGC()}")
```

```python
class Genoma:
  def __init__(self, id, organismo, secuencia):
    self.id = id
    self.organismo= organismo
    self.secuencia= secuencia

  def a_diccionario(self):
    return{
        'id': self.id,
        'organismo': self.organismo,
        'longitud': len(self.secuencia),
        'ContenidoGC': self.secuencia.count('C' + self.secuencia.count('G')/ len(self.secuencia))
    }
```
