```python
def Contenido GC (DNA):
  counter = 0
  if len (DNA) == 3: #Aquí esta integrado el primer IF
   if DNA (0) == 'C' or DNA (0) == 'G'
   counter  = counter + 1
   if DNA (1) == 'C' or DNA (1) == 'G'
   counter  = counter + 1
   if DNA (2) == 'C' or DNA (2) == 'G'
   counter + 1
   return counter /3.0
  elif len (DNA) == 4: #Aquí está ingegrado mi elif
    if DNA (0) == 'C' or DNA (0) == 'G':
      counter = counter + 1
    if DNA (1) == 'C' or DNA (1) == 'G':
      counter = counter + 1
    if DNA (2) == 'C' or DNA (2) == 'G':
      counter = counter + 1
    if DNA (3) == 'C' or DNA(3) == 'G':
      counter == counter +1
    return counter /4.0
    elif len(DNA) == 5:
     if DNA (0) == 'C' or DNA (0) == 'G':
      counter = counter + 1
    if DNA (1) == 'C' or DNA (1) == 'G':
      counter = counter + 1
    if DNA (2) == 'C' or DNA (2) == 'G':
      counter = counter + 1
    if DNA (3) == 'C' or DNA(3) == 'G':
      counter == counter +1
    if DNA (4) == 'C' or DNA (4) == 'G':
      counter = counter + 1
    return counter /5.0
  else:
    print("La secuencia no mide entre 3 y 5 pares de bases")
```
