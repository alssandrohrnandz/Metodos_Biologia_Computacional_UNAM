```python
# Once we have installed the libraries, we can create bar plots to visualize the distribution of anime by media type and source.
# Use the following code to create the bar plots:

plt.figure(figsize = (12, 6)) # In order to set the size of the figure, we use the figsize parameter. The first value is the width and the second is the height
plt.subplot(1, 2, 1) # This line creates a subplot with 1 row and 2 columns, and we are currently working on the first subplot
sns.countplot(data = df_anime_cleaned, x = 'media_type', order = df_anime_cleaned['media_type'].value_counts().index)
plt.xticks(rotation = 45) # This line rotates the x-axis labels by 45 degrees to make them more readable
plt.title('Distribution of Anime by Media Type')
plt.xlabel('Media Type')
plt.ylabel('Count')


# Now we can create the second subplot for the distribution of anime by source.
plt.subplot(1, 2, 2)
sns.countplot(data = df_anime_cleaned, x = 'source', order = df_anime_cleaned['source'].value_counts().index)
plt.xticks(rotation = 90) # This line rotates the x-axis labels by 90 degrees to make them more readable
plt.title('Distribution of Anime by Source')
plt.xlabel('Source')
plt.ylabel('Count')

# Finally, we can display the plots using plt.show()
plt.tight_layout() # This line adjusts the spacing between the subplots to prevent overlapping
plt.show()
```

```python
# ==========================================
# ACTIVITY 3: Filtering and Selection
# ==========================================
# Exercise 1: Filter TV shows with a score greater than 8.0
top_tv_shows = df_anime_clean[(df_anime_clean['media_type'] == 'tv') & (df_anime_clean['mean'] > 8.0)]
print(f"\nNumber of top TV shows (score > 8.0): {len(top_tv_shows)}")

# Exercise 2: Using the .query() method for cleaner syntax
# Find highly popular movies
popular_movies = df_anime_clean.query("media_type == 'movie' and popularity < 500")
```

```python
# ==========================================
# ACTIVITY 4: Groupby and Aggregation
# ==========================================
# Group by media type and calculate the average number of episodes and average score
# Here, we are using the .agg() method to apply multiple aggregation functions to different columns. For example:
# df.groupby('categoria').agg({
#    'precio': 'mean',
#    'ventas': 'sum'
#})

media_stats = df_anime_clean.groupby('media_type').agg({
    'num_episodes': 'mean',
    'mean': 'mean',
    'id': 'count' # Counting the number of anime per media type
}).rename(columns={'id': 'total_count'}) # renaming the 'id' column to 'total_count' for clarity

print("\n--- Statistics by Media Type ---")
display(media_stats.sort_values(by='total_count', ascending=False))
```

```python
# ==========================================
# ACTIVITY 5: Feature Engineering (Dates and Strings)
# ==========================================
# Convert 'start_date' and 'end_date' to datetime objects
# 'errors="coerce"' will turn invalid parsing into NaT (Not a Time)
df_anime_clean['start_date'] = pd.to_datetime(df_anime_clean['start_date'], errors='coerce')
df_anime_clean['end_date'] = pd.to_datetime(df_anime_clean['end_date'], errors='coerce')

# Calculate broadcast duration in days
df_anime_clean['broadcast_duration_days'] = (df_anime_clean['end_date'] - df_anime_clean['start_date']).dt.days

# String manipulation: Create a boolean column if the title contains the word "Evangelion"
# case=False makes it case-insensitive
df_anime_clean['is_evangelion'] = df_anime_clean['title'].str.contains('Evangelion', case=False, na=False)
```

```python
# ==========================================
# ACTIVITY 6: Data Visualization (Plotting)
# ==========================================
# Set seaborn style for better aesthetics
sns.set_theme(style="whitegrid")

# Plot 1: Bar Plot (Pandas built-in)
# Shows the count of anime per media type
plt.figure(figsize=(10, 5))
media_counts = df_anime_clean['media_type'].value_counts()
media_counts.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Number of Anime by Media Type')
plt.xlabel('Media Type')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout() # Adjusts layout so labels don't get cut off
plt.show()
```

```python
# Plot 2: Histogram (Seaborn)
# Distribution of anime scores (mean)
plt.figure(figsize=(8, 5))
sns.histplot(data=df_anime_clean, x='mean', bins=30, kde=True, color='purple')
plt.title('Distribution of Anime Scores')
plt.xlabel('Score (Mean)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plot 3: Scatter Plot (Seaborn)
# Relationship between number of scoring users and the mean score
plt.figure(figsize=(10, 6))
# Using a sample to avoid overplotting if the dataset is too large
sample_df = df_anime_clean.sample(n=2000, random_state=42) 
sns.scatterplot(data=sample_df, x='num_scoring_users', y='mean', alpha=0.6, hue='media_type')
plt.title('Score vs Number of Scoring Users (Sample)')
plt.xlabel('Number of Scoring Users')
plt.ylabel('Score (Mean)')
plt.xscale('log') # Log scale is useful for data with large ranges
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# EXTRA ACTIVITY: Correlation & Outlier Detection (Hidden Gems)
# ==========================================

# 1. Clean data for analysis (Drop rows where 'mean' or 'num_scoring_users' are null)
df_analysis = df_anime.dropna(subset=['mean', 'num_scoring_users']).copy()

# 2. Calculate Pearson correlation between score and number of scoring users
# This answers: Does having more viewers correlate with higher scores?
correlation = df_analysis['mean'].corr(df_analysis['num_scoring_users'])
print(f"\nPearson correlation between Score and Users: {correlation:.2f}")
# A positive correlation usually exists, but it's rarely perfect (1.0).
```

```python
# 3. Find the "Hidden Gems" (Outlier detection)
# Criteria: Very high score (> 8.5) but relatively low number of scoring users (< 10,000)
hidden_gems = df_analysis[(df_analysis['mean'] > 8.5) & (df_analysis['num_scoring_users'] < 10000)]

# Sort them by score to see the absolute best unknown animes
hidden_gems = hidden_gems.sort_values(by='mean', ascending=False)

print(f"\nFound {len(hidden_gems)} 'Hidden Gems' (Score > 8.5, Users < 10,000)")
print("--- Top 5 Hidden Gems ---")
display(hidden_gems[['title', 'media_type', 'mean', 'num_scoring_users']].head())
```

```python
# 4. Visualize the correlation and highlight our hidden gems
plt.figure(figsize=(10, 6))
sns.set_theme(style="darkgrid")

# Plot all data in light gray
sns.scatterplot(
    data=df_analysis, 
    x='num_scoring_users', 
    y='mean', 
    alpha=0.3, 
    color='gray', 
    label='All Anime'
)

# Overlay the hidden gems in bright red
sns.scatterplot(
    data=hidden_gems, 
    x='num_scoring_users', 
    y='mean', 
    color='red', 
    s=70, # Make the dots slightly larger
    label='Hidden Gems',
    edgecolor='black'
)

# Add a regression line to show the general trend
sns.regplot(
    data=df_analysis, 
    x='num_scoring_users', 
    y='mean', 
    scatter=False, 
    color='blue', 
    line_kws={"linewidth": 2, "label": "Trend Line"}
)

plt.title('Anime Score vs Popularity (Finding Hidden Gems)')
plt.xlabel('Number of Scoring Users (Popularity)')
plt.ylabel('Score (Mean)')
plt.xscale('log') # Log scale is essential here due to huge popularity differences
plt.legend()
plt.tight_layout()
plt.show()
```
