# Git Core Concepts: Push vs. Pull & Saving to GitHub

Before uploading your work, it is essential to understand how Google Colab communicates with your GitHub repository.

## Understanding Pull and Push: The Mental Model

Think of GitHub as a central cloud storage server (a "remote repository") and your local workspace (Google Colab or VS Code) as your working desk.

* Pull (Retrieving Changes):
    * What it does: Downloads the latest version of the code and files from GitHub into your workspace.
    * When to use it: Always do a Pull before you start working to ensure you have the most up-to-date version, especially when working in teams or across multiple devices.
    * Analogy: Fetching the latest copy of a shared document from the cloud so you don't edit an outdated version.

![Pull](../Imagenes/PullGitHub.jpg)

* Push (Uploading Changes):
    * What it does: Sends your local changes, new files, and saved commits back up to GitHub.
    * When to use it: Do a Push whenever you complete a logical piece of work or finish your assignment.
    * Analogy: Uploading your finished essay to Google Drive so your teacher can grade it.

![Push](../Imagenes/PushGitHub.jpg)

## Step-by-Step: Uploading Your Colab Notebook to GitHub

### Step 1: Prepare Your Notebook

1) Make sure your notebook has the **Standard Documentation Header** in the first text cell (_Name_, _Date_, _Objectives_, etc.).
2) Run all code cells (`Runtime → Run all` or `Ctrl+Enter`) to confirm there are no syntax errors and that all outputs (e.g., plots, print statements) are displayed.

### Step 2: Push Your Notebook to GitHub

1) In the top navigation bar of Google Colab, click File.
2) Select **Save a copy in GitHub...**
![Save in GitHub option](../Imagenes/Save_Copy_GitHub.png)
![Link to GitHub](../Imagenes/LinkToGitHub.png)
3) In the pop-up configuration window:
    * **Repository**: Choose your repository from the drop-down menu (`My_First_Repository_YOURINITIALS`).
    * **Branch**: Select `main`.
    * **FilePath**: Specify the file name (e.g., `homework_01.ipynb` or `notebooks/homework_01.ipynb`).
    * **Commit message**: Write a clear, concise description of your changes (e.g., `Add homework 1 with spatial data analysis`).
    * Include a link to **Open in Colab**: Ensure this checkbox is checked ✓.
    ![Options activated](./Imagenes/GitHubOptions.png)
4) Click **OK**.

### Step 3: Confirm your upload

1) Open a new tab and go to your GitHub repository URL:
`https://github.com/your-username/My_First_Repository_YOURINITIALS`

# Assignment 1: How to Complete and Submit Assignments Using GitHub Fork & Pull Requests

Follow these step-by-step instructions to create a fork of the assignment repository, complete your notebook, and submit your work for grading.

1) Navigate to the official assignment repository URL provided by your instructor `(e.g., https://github.com/instructor-username/Assignment_01).`
2) In the top-right corner of the GitHub page, click the **Fork** button (or click Create a new fork).
3) Under Owner, select your GitHub username.
4) Keep the repository name as is (or adjust if instructed), leave "**Copy the main branch only**" checked ✓, and click **Create fork**.
5) You now have your own independent copy of the assignment at `https://github.com/your-username/Assignment_01!`

## Step 2: Open and Complete the Notebook in Google Colab

1) Go to [Colab](colab.research.google.com)
2) Select **File → Open notebook** and click the GitHub tab.
3) Select YOUR username and choose your forked repository (`Assignment_01`).
4) Click on the assignment notebook file (e.g., `assignment_01.ipynb`).
5) Ensure Include a link to **Open in Colab** is checked ✓ and click **OK**.

## Step 3: Push Your Changes Back to Your Fork

1) Once you finish your assignment in Colab, go to **File → Save** a copy in GitHub...
2) Ensure your repository is selected (`your-username/Assignment_01`).
3) Set the Branch to main.
4) Write a descriptive commit message (e.g., ***Completed assignment 1 on population genetics***).
5) Ensure Include a link to **Open in Colab** is checked ✓ and click **OK**.

## Submit Your Homework via Pull Request (PR)

1) Go to your forked repository on GitHub (`https://github.com/your-username/Assignment_01`).
2) You will see a banner near the top saying: ***"This branch is 1 commit ahead of instructor-username:main"***.
3) Click the **Contribute** button next to it and select **Open pull request**.
4) Review the differences to ensure your completed notebook is included.
5) Title your Pull Request with your full name and student ID (e.g., `Submission: Assignment 1 - Jessica Simpson`)
6) In the description box, write a brief note for your instructor detailing what you completed.
7) Click the green ***Create pull request*** button.
