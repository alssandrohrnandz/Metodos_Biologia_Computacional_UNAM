# How biologists save data?

A data frame (in `pandas`, `pd.DataFrame()`) is a two-dimensional, tabular data structure structured with rows and columns. Unlike a matrix, a data frame can store columns containing different data types (e.g., *numbers*, *strings*, *dates*).

![A Pandas DataFrame](../Imagenes/Pandas_DF.PNG)

To create our first dataframe we use the function `pd.DataFrame()` as following:

```python
import pandas as pd

# initialize data of lists.
data = {'Name': ['Tom', 'nick', 'krish', 'jack'],
        'Age': [20, 21, 19, 18]}

# Create DataFrame
df = pd.DataFrame(data)

print(df)
```

In Python (`pandas`), a DataFrame is an assembly of `Series` aligned by a shared index. While dictionaries with **key-value pairs** are a common way to construct them using column labels, pandas can also build DataFrames from `NumPy` arrays, lists of records, or external files. The following are examples of contruction of DF using lists, lists of dictionaries, another dataframe and series. 

```python
# 1. Using Lists (List of Lists)
data_lists = [
    ['Ana', 25, 'Madrid'],
    ['Luis', 30, 'Barcelona'],
    ['Carla', 22, 'Valencia']
]
df_from_lists = pd.DataFrame(data_lists, columns=['Name', 'Age', 'City'])

# 2. Using Lists of Dictionaries
data_dicts = [
    {'Name': 'Ana', 'Age': 25, 'City': 'Madrid'},
    {'Name': 'Luis', 'Age': 30, 'City': 'Barcelona'},
    {'Name': 'Carla', 'Age': 22, 'City': 'Valencia'}
]
df_from_dicts = pd.DataFrame(data_dicts)

# 3. Using Series (Dictionary of Series)
s_name = pd.Series(['Ana', 'Luis', 'Carla'])
s_age = pd.Series([25, 30, 22])
s_city = pd.Series(['Madrid', 'Barcelona', 'Valencia'])

df_from_series = pd.DataFrame({
    'Name': s_name,
    'Age': s_age,
    'City': s_city
})

# 4. Using Another DataFrame (Copying / Filtering)
df_from_df = df_from_lists.copy()
```

As you have observed, dictionaries are key-value mappings that allow us to organize and label structured data. For example:

```python
# 1. Basic Dictionary (Representing a single entity)
person = {
    "name": "Ana",
    "age": 25,
    "city": "Madrid"
}

# Accessing values by key
print(person["name"])  # Output: Ana
```

```python
# 2. Dictionary of Lists (Ideal structure to build a pandas DataFrame)
data_dict = {
    "Name": ["Ana", "Luis", "Carla"],
    "Age": [25, 30, 22],
    "City": ["Madrid", "Barcelona", "Valencia"]
}

# Accessing a full column/list by key
print(data_dict["Age"])  # Output: [25, 30, 22]
```

This is very useful for biologists, given the vast amount of biological data available. For example:

```python
# Organizing gene expression data using a dictionary of lists
gene_data = {
    "Gene_ID": ["BRCA1", "TP53", "EGFR"],
    "Sample_A_Expression": [12.5, 45.2, 8.1],
    "Sample_B_Expression": [14.1, 41.8, 12.3],
    "Is_Mutated": [True, True, False]
}

# Converting biological data into a DataFrame for analysis
df_genes = pd.DataFrame(gene_data)

print(df_genes)
```

```python
# Solution

drug_experiment = {
    "Well": ["A1", "A2", "B1", "B2"],
    "Treatment_uM": [0.0, 0.0, 10.0, 10.0],
    "Viability_Percent": [98.5, 96.2, 45.1, 48.3]
}

df_experiment = pd.DataFrame(drug_experiment)
print(df_experiment)

# Optional Solution

df_experiment_filtered= df_experiment[df_experiment["Treatment_uM"]  > 0]
display(df_experiment_filtered)
```

Also, you can group the data by `Treatment_uM` and calculate the average `Viability_Percent` for control vs. treated samples.

```python
mean_viability = df_experiment.groupby("Treatment_uM")["Viability_Percent"].mean()

print(mean_viability)
```

1) `df_experiment.groupby("Treatment_uM")`: divide the rows into groups based on their values in `Treatment_uM`column (one group of $0.0$ and another group for $10.0$)
2) `["Viability_Percent"]`: select only the viability column on which we want to calculate the `mean()`
3) `.mean()`: apply the mean over each resulting group

If you only want to keep the result as a new DF (instead of a panda `Series`), we can add doble square brackets `[["Viability_Percent"]]` or use `.reset_index()` 

```python
# Alternative: Output as a clean DataFrame
mean_viability_df = df_experiment.groupby("Treatment_uM")[["Viability_Percent"]].mean().reset_index()
print(mean_viability_df)
```

### Explanation of `print(f"Redundancy factor (sites): {rows / unique_sites:.1f}x")`

* `f"..."` → This is an f-string, which allows you to embed variables or expressions directly inside a string.
* `rows / unique_sites` → This calculates the ratio between the total number of rows (data entries) and the number of unique sites. **In other words, it measures how many times, on average, each site appears in the dataset.**
* `:.1f` → This formatting instruction tells Python to display the result as a floating-point number with one decimal place.
* `x` → The `x` at the end is just literal text, added to indicate a multiplication factor (like “2.5x”).
* `print(...)` → Finally, it outputs the formatted string to the console.

```python
import os
import numpy as np
import pandas as pd

# Reproducibilidad: los CSV deben ser siempre idénticos
SEED = 42
np.random.seed(SEED)

# Carpeta de salida
OUT = "../input/tu_nombre"
os.makedirs(OUT, exist_ok=True)


def guardar(df, nombre, tarea):
    """Guarda un DataFrame en input/ y reporta el resultado."""
    ruta = os.path.join(OUT, nombre)
    df.to_csv(ruta, index=False, encoding="utf-8")
    print(f"  ✓ [{tarea:>8}] {ruta}   ({len(df)} filas × {len(df.columns)} cols)")


print("Generando archivos CSV de práctica...\n")


# ======================================================================
# TASK 1 — Loading and inspecting
# ======================================================================
# Dataset limpio y pequeño. Sirve para practicar:
#   read_csv, head, info, dtypes, shape, columns, isna().sum(), describe

task1 = pd.DataFrame({
    "id_muestra":  ["M001", "M002", "M003", "M004", "M005"],
    "sitio":       ["El Roble", "El Roble", "La Campana", "La Campana", "Senda Darwin"],
    "especie":     ["Liolaemus tenuis", "Rhinella spinulosa", "Liolaemus tenuis",
                    "Pleurodema thaul", "Alsodes nodosus"],
    "peso_g":      [5.2, 12.3, 4.8, 3.1, 6.7],
})
guardar(task1, "task1_general.csv", "Task 1")


# ======================================================================
# TASK 2 — Duplicates
# ======================================================================
# Contiene:
#   - 2 filas EXACTAMENTE duplicadas (M002 aparece dos veces idéntica)
#   - 2 filas "casi" duplicadas (M004: mismo id, sitio, especie y peso,
#     pero distinta longitud)
# Sirve para: duplicated(), duplicated(keep=...), drop_duplicates()

task2 = pd.DataFrame({
    "id_muestra":  ["M001", "M002", "M002", "M003", "M004", "M004"],
    "sitio":       ["El Roble", "El Roble", "El Roble", "La Campana",
                    "La Campana", "La Campana"],
    "especie":     ["Liolaemus tenuis", "Rhinella spinulosa", "Rhinella spinulosa",
                    "Liolaemus tenuis", "Liolaemus tenuis", "Liolaemus tenuis"],
    "peso_g":      [5.2, 12.3, 12.3, 4.8, 5.5, 5.5],
    "longitud_cm": [6.1, 7.4, 7.4, 5.9, 6.3, 6.5],
})
guardar(task2, "task2_duplicates.csv", "Task 2")


# ======================================================================
# TASK 3 — Text standardization
# ======================================================================
# Nombres de sitio con:
#   - espacios al inicio/final
#   - espacios dobles
#   - mayúsculas inconsistentes
#   - variantes de "el" vs "El"
# Sirve para: strip(), lower(), split(), " ".join(), apply()

task3 = pd.DataFrame({
    "id_muestra": ["M001", "M002", "M003", "M004", "M005", "M006"],
    "sitio": [
        "  Reserva El Roble ",
        "reserva  el roble",
        "RESERVA EL ROBLE",
        "Reserva el Roble ",
        "Parque  Nacional La Campana",
        "parque nacional la campana",
    ],
    "peso_g": [5.2, 4.8, 5.5, 6.0, 7.1, 6.9],
})
guardar(task3, "task3_messy_text.csv", "Task 3")


# ======================================================================
# TASK 4 — Mapping with a dictionary
# ======================================================================
# Códigos de especie que deben mapearse a nombres completos.
# Se incluye un código DESCONOCIDO ("XXX") que no está en el diccionario
# para que el estudiante observe qué hace .map() en ese caso (devuelve NaN).

task4 = pd.DataFrame({
    "id_muestra":     ["M001", "M002", "M003", "M004", "M005", "M006"],
    "codigo_especie": ["LTE", "LCH", "RSP", "LTE", "PTH", "XXX"],
    "peso_g":         [5.2, 6.8, 12.3, 4.9, 3.1, 5.0],
})
guardar(task4, "task4_species_codes.csv", "Task 4")


# ======================================================================
# TASK 5 — Missing values
# ======================================================================
# NaN distribuidos en distintas columnas, para comparar:
#   dropna()                      → elimina cualquier fila con NaN
#   dropna(subset=["peso_g"])     → elimina solo si falta peso_g
#   fillna(...)                   → rellena
# También sirve para isna().sum() por columna.

task5 = pd.DataFrame({
    "id_muestra":  ["M001", "M002", "M003", "M004", "M005", "M006"],
    "sitio":       ["El Roble", None, "La Campana", "La Campana", None, "Senda Darwin"],
    "especie":     ["Liolaemus tenuis", "Rhinella spinulosa", None,
                    "Liolaemus tenuis", "Pleurodema thaul", "Alsodes nodosus"],
    "peso_g":      [5.2, 12.3, 4.8, None, 3.1, 6.7],
    "longitud_cm": [6.1, None, 5.9, 6.3, 4.5, None],
})
guardar(task5, "task5_missing.csv", "Task 5")


# ======================================================================
# TASK 7 — Boolean filtering
# ======================================================================
# Valores numéricos limpios, para practicar:
#   df[df["col"] > x]
#   df[(df["a"] > x) & (df["b"] < y)]
#   df[~mask]
# El dataset está pensado para que el filtro
# "peso_g > 5 AND longitud_cm < 8" deje solo algunas filas.

task7 = pd.DataFrame({
    "id_muestra":  ["M001", "M002", "M003", "M004", "M005", "M006", "M007", "M008"],
    "especie":     ["L. tenuis", "L. tenuis", "R. spinulosa", "R. spinulosa",
                    "P. thaul", "P. thaul", "A. nodosus", "A. nodosus"],
    "peso_g":      [4.5, 5.8, 11.2, 13.5, 2.8, 3.5, 6.9, 7.4],
    "longitud_cm": [5.9, 6.4, 7.2, 7.8, 4.1, 4.4, 6.8, 7.1],
})
guardar(task7, "task7_filtering.csv", "Task 7")


# ======================================================================
# TASK 8 — Grouping and aggregation
# ======================================================================
# 15 filas repartidas en 4 especies con tamaños de grupo distintos
# (5, 4, 3, 3). Sirve para practicar groupby().agg(), value_counts(),
# nunique(), reset_index(), sort_values().
# Además hay una columna "sitio" para poder agrupar por dos variables.

task8 = pd.DataFrame({
    "id_muestra": [f"M{i:03d}" for i in range(1, 16)],
    "especie": (
        ["L. tenuis"] * 5 +
        ["R. spinulosa"] * 4 +
        ["P. thaul"] * 3 +
        ["A. nodosus"] * 3
    ),
    "sitio": [
        "El Roble", "El Roble", "La Campana", "La Campana", "Senda Darwin",
        "El Roble", "El Roble", "La Campana", "Senda Darwin",
        "El Roble", "La Campana", "Senda Darwin",
        "El Roble", "La Campana", "Senda Darwin",
    ],
    "peso_g": [
        5.2, 4.8, 5.5, 6.1, 4.9,
        12.3, 11.8, 13.5, 12.0,
        3.1, 2.9, 3.4,
        6.7, 7.2, 6.5,
    ],
})
guardar(task8, "task8_groups.csv", "Task 8")


# ======================================================================
# TASK 11 — Correlation
# ======================================================================
# Dos especies con 20 individuos cada una:
#   Especie A → correlación peso–longitud FUERTE (cercana a 1)
#   Especie B → correlación peso–longitud DÉBIL (cercana a 0)
#
# Sirve para que descubran que la correlación global puede diferir
# de la correlación POR GRUPO (antesala de la paradoja de Simpson).

n = 20

# Especie A: peso = 0.8 * longitud + ruido pequeño → correlación alta
length_a = np.random.uniform(5, 10, n)
weight_a = 0.8 * length_a + np.random.normal(0, 0.3, n)

# Especie B: peso independiente de longitud → correlación baja
length_b = np.random.uniform(5, 10, n)
weight_b = np.random.normal(7, 2, n)

task11 = pd.DataFrame({
    "id_muestra":  ([f"A{i:03d}" for i in range(1, n + 1)] +
                    [f"B{i:03d}" for i in range(1, n + 1)]),
    "especie":     ["Especie A"] * n + ["Especie B"] * n,
    "longitud_cm": np.round(np.concatenate([length_a, length_b]), 2),
    "peso_g":      np.round(np.concatenate([weight_a, weight_b]), 2),
})
guardar(task11, "task11_correlation.csv", "Task 11")


# ======================================================================
# TASK 12 — First plot
# ======================================================================
# Dataset pensado para dos gráficos:
#   1) Barh: número de muestras por sitio (3 sitios con 5, 4 y 3 muestras)
#   2) Scatter: longitud_cm vs peso_g (dos "nubes" de puntos separadas)
# Sirve para: plot(kind="barh"), scatter, savefig, plt.close(fig)

task12 = pd.DataFrame({
    "id_muestra":  [f"M{i:03d}" for i in range(1, 13)],
    "sitio":       ["El Roble"] * 5 + ["La Campana"] * 4 + ["Senda Darwin"] * 3,
    "longitud_cm": [6.1, 5.9, 6.4, 5.8, 6.2,
                    7.2, 7.5, 6.9, 7.4,
                    4.5, 4.8, 4.2],
    "peso_g":      [5.2, 4.8, 5.5, 4.9, 5.3,
                    12.1, 13.2, 11.5, 12.4,
                    3.2, 3.5, 2.9],
})
guardar(task12, "task12_plotting.csv", "Task 12")


# ======================================================================
# Resumen final
# ======================================================================

print("\n" + "=" * 60)
print("LISTO. Archivos generados en la carpeta 'input/':")
print("=" * 60)
print("""
  task1_general.csv         → Task 1  (inspección básica)
  task2_duplicates.csv      → Task 2  (duplicados exactos y parciales)
  task3_messy_text.csv      → Task 3  (estandarización de texto)
  task4_species_codes.csv   → Task 4  (mapa con diccionario)
  task5_missing.csv         → Task 5  (valores faltantes)
  task7_filtering.csv       → Task 7  (filtros booleanos)
  task8_groups.csv          → Task 8  (groupby + agg)
  task11_correlation.csv    → Task 11 (correlación por grupo)
  task12_plotting.csv       → Task 12 (primeros gráficos)
""")
print("Cada estudiante debe cargar el CSV de su tarea con pd.read_csv('input/<archivo>.csv').")
```
