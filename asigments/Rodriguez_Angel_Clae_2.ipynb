```python
def ContenidoGC(DNA):
  counter = 0
  if len(DNA) == 3: #el primer if
    if DNA[0] =='C' or DNA[0] == 'G':
      counter = counter + 1
    if DNA[1] =='C' or DNA[1] == 'G':
      counter = counter + 1
    if DNA[2] =='C' or DNA[2] == 'G':
      counter = counter + 1
    return counter / 3.0
  elif len(DNA) == 4:
    if DNA[0] =='C' or DNA[0] == 'G':
      counter = counter + 1
    if DNA[1] =='C' or DNA[1] == 'G':
      counter = counter + 1
    if DNA[2] =='C' or DNA[2] == 'G':
      counter = counter + 1
    if DNA[3] =='C' or DNA[3] == 'G':
      counter = counter + 1
    return counter / 4.0
  else:
    if DNA[0] =='C' or DNA[0] == 'G':
      counter = counter + 1
    if DNA[1] =='C' or DNA[1] == 'G':
      counter = counter + 1
    if DNA[2] =='C' or DNA[2] == 'G':
      counter = counter + 1
    if DNA[3] =='C' or DNA[3] == 'G':
      counter = counter + 1
    if DNA[4] == 'C' or DNA[4] == 'G':
      counter = counter + 1
  return counter / 5.0
```

```python
#LOOPS
seq_3='ATGCGCGTAGCACGTAGCGATCGATCGATSGCTCGAACGAT6AGCTAGCTAGCTAGCTCAGATCGTGT'
```

```python
len(seq_3)
```

```python
#La funcion 'FOR' va a evaluar sobre cada posicion de la secuncia
def ContenidoGC_v2(DNA):
  counter = 0
  for base in DNA:
    if base == 'C' or base == 'G':
      counter = counter + 1
  return counter / len(DNA)
```

```python
ContenidoGC_v2(seq_3)
```
